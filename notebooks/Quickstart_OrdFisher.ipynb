# ordfisher 1.0.0: reproducible quickstart

This notebook uses the **locally installed** package. In Colab, upload or clone the *real* version 1.0.0 source repository into the runtime first; then run `%pip install .` from its root. There is no public repository URL in this candidate, and none is assumed.

All spectra use lexicographic `argsort` patterns and the adjacent-rank Cayley graph. The examples here are brief demonstrations, **not** the full 510-trajectory logistic experiment.


```python
import ordfisher, numpy as np
print("ordfisher:", ordfisher.__version__)
assert ordfisher.__version__ == "1.0.0"
```

## 1. OPS and OFS from a time series

The numerical counting backend uses stable sorting of exact ties by time index. The graph frequencies are not temporal Fourier frequencies.


```python
from ordfisher import spectra_from_series, check_spectral_identities
x = np.random.default_rng(42).normal(size=10000)
r = spectra_from_series(x, m=4, tau=1)
print("Cayley Fisher energy:", r.fisher_energy)
print("OPS/OFS normalized-profile distance:", r.ops_ofs_tv())
print("Identities:", check_spectral_identities(r.probabilities, m=4))
```

## 2. Recreate the appendix Cayley graph

The fixed layout and visual conventions correspond to the manuscript's S4 appendix figure. Font rasterization may vary between systems.


```python
import matplotlib.pyplot as plt
from ordfisher import cayley_graph, plot_cayley_graph
G = cayley_graph(4)
print('S4:', G.number_of_nodes(), 'vertices,', G.number_of_edges(), 'edges')
fig, ax = plot_cayley_graph(4, layout="appendix")
plt.show()
```

## 3. Exact rational Markov certificate

This is a *newly reconstructed* exact rational certificate of the same stated mathematical constraints. It is not the unrecovered original file.


```python
from ordfisher.markov import verify_exact_certificate
report = verify_exact_certificate()
print("Exact ordinal residuals:", report['plus']['max_exact_ordinal_residual'], report['minus']['max_exact_ordinal_residual'])
print("OFS normalized-profile TV:", report['spectral_numerical']['tv_ofs_normalized'])
```
